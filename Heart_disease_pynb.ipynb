#data pre prossing and cleaning

```python
df_encode = pd.get_dummies(df,columns=['Sex','ChestPainType','RestingECG','ExerciseAngina','ST_Slope'])
```

```python
df_encode
```

```python
df_encode = df_encode.astype(int)
```

```python
df_encode
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
numerical_columns = ['Age','RestingBP','Cholesterol','MaxHR','Oldpeak']
scaler = StandardScaler()
```

```python
df_encode[numerical_columns] = scaler.fit_transform(df_encode[numerical_columns])
```

```python
df_encode
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix,f1_score,classification_report
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
x= df_encode.drop('HeartDisease', axis=1)
y= df_encode['HeartDisease']
```

```python
x
```

```python
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(
   x, y, test_size=0.23, random_state=42)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.fit_transform(x_test)
```

```python
model ={
    "LogisticRegression": LogisticRegression(),
    "DecisionTreeClassifier": DecisionTreeClassifier(),
    "SVM(RBF kernel)": SVC(),
    "KNeighborsClassifier": KNeighborsClassifier(),
    "Naive Bayes": GaussianNB()
}
```

```python
result =[]
```

```python
model
```

```python
for  name ,current_model in model.items():
    current_model.fit(x_train_scaled,y_train)
    y_pred = current_model.predict(x_test_scaled)
    acc = accuracy_score(y_test,y_pred)
    f1= f1_score(y_test,y_pred)
    result.append({
      'model': name,
      'Accuracy':round(acc, 4),
      'F1 Score':round(f1,4),
     })
```

```python
results= pd.DataFrame(result)
```

```python
import joblib
import numpy as np
import pandas as pd
import gradio as gr

# Load files
with open("LogisticRegression_heart.pkl", "rb") as f:
    model = joblib.load(f)

with open("scaler.pkl", "rb") as f:
    scaler = joblib.load(f)

with open("columns.pkl", "rb") as f:
    columns = joblib.load(f)

# Prediction Function
def predict_heart(*inputs):

    input_df = pd.DataFrame([inputs], columns=columns)

    scaled_input = scaler.transform(input_df)

    prediction = model.predict(scaled_input)[0]

    probability = model.predict_proba(scaled_input)[0]

    if prediction == 1:
        result = "⚠️ Heart Disease Detected"
    else:
        result = "✅ No Heart Disease Detected"

    confidence = max(probability) * 100

    return f"{result}\nConfidence: {confidence:.2f}%"

# Create input components automatically
inputs = []

for col in columns:
    inputs.append(
        gr.Number(label=col)
    )

# Gradio Interface
interface = gr.Interface(
    fn=predict_heart,
    inputs=inputs,
    outputs=gr.Textbox(label="Prediction"),
    title="Heart Disease Prediction",
    description="Enter patient details to predict heart disease."
)

interface.launch(share=True)
```
