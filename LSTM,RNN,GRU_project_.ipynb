```python
index_to_word = {}
for word, index in word_index.items():
  index_to_word[index] = word
```

```python
index_to_word
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
def predictor(model,Tokenizer,text,next_words=1):
  text=text.lower()

  seq=Tokenizer.texts_to_sequences([text])[0]
  for _ in range(next_words):
    seq=pad_sequences([seq],maxlen=max_len,padding='pre')
  pred =model.predict(seq)
  pred_index =np.argmax(pred)
  predicted_word = index_to_word[pred_index]
  return predicted_word ,pred_index
```

```python
seed_text ="who"
next_word =predictor(rnn_model_1,tok,seed_text,next_words=1)
print(next_word)
```

```python
def generate_text(model, Tokenizer, seed_text, max_len, num_words_to_generate):
   generated_text = seed_text
   for _ in range(num_words_to_generate):
     pred_word, pred_index = predictor(model, Tokenizer, generated_text, next_words=1)
     # Stop if prediction is empty or some other stopping condition if needed
     if pred_word == "":
        break
     generated_text += " " + pred_word
   return generated_text
```

```python
seed_text="the meaning of life "
generated_output_text = generate_text(rnn_model_1, tok, seed_text, max_len, num_words_to_generate=10)
print(generated_output_text)
```
